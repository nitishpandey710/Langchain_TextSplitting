```python
from langchain.text_splitter import RecursiveCharacterTextSplitter,Language
```

```python
text="""# Introduction

Artificial Intelligence (AI) is changing the world. It powers applications from chatbots to autonomous vehicles.

## History

The field of AI began in the 1950s. Since then, it has gone through several cycles of growth and setbacks.

## Applications

AI is used in various domains:

### Healthcare

AI helps in diagnosis, drug discovery, and personalized treatment.

### Finance

It is used for fraud detection, risk assessment, and algorithmic trading.

# Conclusion

AI continues to evolve, promising even more breakthroughs in the years ahead.
"""
```

```python
# Initialize the splitter
splitter = RecursiveCharacterTextSplitter.from_language(
    language=Language.MARKDOWN,
    chunk_size=200,
    chunk_overlap=0,
)
```

```python
splitter
```

```python
#split
chunks = splitter.split_text(text)
```

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter
```

```python
text = """
Mike Tyson, often called "The Baddest Man on the Planet," became the youngest heavyweight boxing champion at just 20 years old. Known for his explosive power and aggressive style, Tyson dominated the boxing world in the late 1980s. His speed, strength, and intimidation made him a fearsome opponent in the ring.
Beyond boxing, Tyson's life has been filled with both triumphs and controversies. Despite personal struggles, he remains a cultural icon, inspiring generations through his comeback stories, interviews, and appearances in movies and shows.
"""
```

```python
# Initialize the splitter
splitter = RecursiveCharacterTextSplitter(
    chunk_size=200,
    chunk_overlap=0,
)
```

```python
splitter
```

```python
# Perform the split
chunks = splitter.split_text(text)
```
